### 3.2 Feature Selection and Model Training

The next step is to prepare the variables for Linear Regression. The categorical variables are converted into numerical variables so that they can be included in the model. The variable **charges** is retained as the target variable, while **age, BMI, children, sex, smoking status and region** are considered predictor variables.

The categorical variables are encoded using dummy variables. In this process, **sex_male** represents male observations, **smoker_yes** represents smokers, and the regional variables represent the relevant geographic categories. This allows the categorical information to be incorporated into the Linear Regression model.

The resulting dataset contains the numerical predictor variables together with the encoded categorical variables. The first five observations are displayed below to confirm that the feature preparation has been completed correctly.

```python
# Q3.2: Encode categorical variables

df_encoded = pd.get_dummies(
    df,
    columns=["sex", "smoker", "region"],
    drop_first=True,
    dtype=int
)

df_encoded.head()
```

```python
# Q3.2: Split the data into training and testing sets

from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression

# Define the target variable
y = df_encoded["charges"]

# Define the predictor variables
X = df_encoded.drop("charges", axis=1)

# Split the data: 80% for training and 20% for testing
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.20, random_state=42
)

print("Training data shape:", X_train.shape)
print("Testing data shape:", X_test.shape)
```

```python
# Q3.2: Train the Linear Regression model

# Create the Linear Regression model
model = LinearRegression()

# Train the model using the training data
model.fit(X_train, y_train)

print("Linear Regression model trained successfully.")
```

```python
# Q3.2: Make predictions on the testing data

y_pred = model.predict(X_test)

print("First 10 actual charges:")
print(y_test.head(10).values)

print("\nFirst 10 predicted charges:")
print(y_pred[:10])
```

```python
# Q3.2: Examine the model coefficients

coefficients = pd.DataFrame({
    "Feature": X.columns,
    "Coefficient": model.coef_
})

coefficients = coefficients.sort_values(
    by="Coefficient",
    ascending=False
)

print("Linear Regression coefficients:")
display(coefficients)
```

## Question 4: Evaluate the Linear Regression Model

The trained Linear Regression model will now be evaluated using the testing dataset. The evaluation will use R-squared (R²), Mean Absolute Error (MAE), Mean Squared Error (MSE) and Root Mean Squared Error (RMSE). These measures will help determine how accurately the model predicts medical insurance charges on unseen data.

```python
# Q4: Evaluate the Linear Regression model

from sklearn.metrics import r2_score, mean_absolute_error, mean_squared_error
import numpy as np

# Calculate evaluation metrics
r2 = r2_score(y_test, y_pred)
mae = mean_absolute_error(y_test, y_pred)
mse = mean_squared_error(y_test, y_pred)
rmse = np.sqrt(mse)

print("Model Evaluation Results")
print("------------------------")
print("R-squared (R²):", round(r2, 4))
print("Mean Absolute Error (MAE):", round(mae, 2))
print("Mean Squared Error (MSE):", round(mse, 2))
print("Root Mean Squared Error (RMSE):", round(rmse, 2))
```

### Interpretation of the Model Evaluation

The Linear Regression model produced an R-squared value of **0.8069**, meaning that approximately **80.69% of the variation in medical insurance charges is explained by the variables included in the model**. This indicates that the model has a reasonably strong predictive relationship with the target variable.

The **Mean Absolute Error (MAE) of 4,177.05** indicates that the model's predictions differ from the actual insurance charges by approximately 4,177.05 charge units on average. The **Root Mean Squared Error (RMSE) of 5,956.34** is higher than the MAE, indicating that some predictions have larger errors than others. This is consistent with the variation and high-value observations identified during the EDA.

The **Mean Squared Error (MSE) of 35,478,020.68** provides another measure of prediction error by giving greater weight to larger errors. Overall, the evaluation results suggest that the model performs reasonably well, although it does not predict every individual's insurance charges accurately.

For the medical aid scenario, the model could therefore provide useful estimates for supporting a sliding scale of charges. ## 4.2 Retrain with Different Parameters

The initial Linear Regression model achieved an R-squared value of 0.8069, indicating reasonably strong predictive performance. However, the residual analysis shows that prediction errors are not completely random, particularly for some customers with higher medical insurance charges. This suggests that the basic linear model may not capture all relationships between the explanatory variables and insurance charges.

To investigate whether the model can be improved, a second Linear Regression model will be developed by introducing interaction and nonlinear features. In particular, an interaction between BMI and smoking status is considered because the exploratory analysis indicated that the relationship between BMI and medical insurance charges may differ according to smoking status. A squared BMI term is also introduced to allow the model to capture a possible nonlinear relationship between BMI and charges.

The improved model will be evaluated on the same testing data using R-squared, MAE and RMSE. Its performance will then be compared with the original model to determine whether the additional features improve prediction.

```python
# Question 4.2: Retrain the model with additional features

# Create copies so that the original datasets remain unchanged
X_train_improved = X_train.copy()
X_test_improved = X_test.copy()

# Add interaction and nonlinear features
X_train_improved["bmi_smoker_interaction"] = (
    X_train_improved["bmi"] * X_train_improved["smoker_yes"]
)
X_test_improved["bmi_smoker_interaction"] = (
    X_test_improved["bmi"] * X_test_improved["smoker_yes"]
)

X_train_improved["bmi_squared"] = X_train_improved["bmi"] ** 2
X_test_improved["bmi_squared"] = X_test_improved["bmi"] ** 2

# Train the improved Linear Regression model
improved_model = LinearRegression()
improved_model.fit(X_train_improved, y_train)

# Make predictions
y_pred_improved = improved_model.predict(X_test_improved)

# Evaluate the improved model
r2_improved = r2_score(y_test, y_pred_improved)
mae_improved = mean_absolute_error(y_test, y_pred_improved)
mse_improved = mean_squared_error(y_test, y_pred_improved)
rmse_improved = np.sqrt(mse_improved)

print("Original Model vs Improved Model")
print("--------------------------------")
print("Original R-squared:", round(r2, 4))
print("Improved R-squared:", round(r2_improved, 4))
print()
print("Original MAE:", round(mae, 2))
print("Improved MAE:", round(mae_improved, 2))
print()
print("Original RMSE:", round(rmse, 2))
print("Improved RMSE:", round(rmse_improved, 2))
```

### Interpretation of the Retrained Model

The retrained Linear Regression model produced a clear improvement over the original model. The R-squared value increased from **0.8069 to 0.8873**, indicating that the improved model explains approximately **88.73% of the variation in medical insurance charges**, compared with 80.69% explained by the original model.

Prediction errors also decreased substantially. The Mean Absolute Error (MAE) decreased from **4,177.05 to 2,822.02**, meaning that the improved model's predictions are, on average, considerably closer to the actual insurance charges. Similarly, the Root Mean Squared Error (RMSE) decreased from **5,956.34 to 4,549.80**, demonstrating that the improved model also reduced the impact of larger prediction errors.

The improvement occurred after introducing the **BMI–smoking interaction** and a **squared BMI term**. This suggests that the relationship between BMI and medical insurance charges is not entirely linear and that the effect of BMI may differ according to smoking status. These additional features therefore allow the Linear Regression model to represent patterns in the dataset that were not adequately captured by the original specification.

Overall, the retrained model is preferred because it achieves a higher R-squared value together with lower MAE and RMSE values on the same unseen testing data. Nevertheless, approximately **11.27% of the variation in charges remains unexplained**, indicating that other relevant factors or more flexible modelling approaches could potentially improve prediction further.
